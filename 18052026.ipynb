```python
import numpy as np
import pandas as pd
import time
```

```python
import ipywidgets as widgets #Подключаем библиотеку виджетов
from IPython.display import display # импортируем функцию для визуализации виджетов
```

```python
Data = pd.read_csv('https://raw.githubusercontent.com/junaart/ForStudents/refs/heads/main/Python/data_1.csv')
Data
```

```python
global  p
p = -1
h = widgets.Text(
    value='',
    description='ФИО:',
    disabled=False
)
b1 = widgets.Button(
    description='>',
    disabled=False
)
b2 = widgets.Button(
    description='<',
    disabled=False
)

d1 = widgets.DatePicker(
    description='Дата рождения: ',
    disabled=False
)

d2 = widgets.DatePicker(
    description='Дата приема на работу: ',
    disabled=False
)
s = widgets.RadioButtons(
    options=['м','ж'],
    description='Пол:',
    disabled=False
)

oklad = widgets.IntText(
    description='Оклад: ',
    value = 0
)

family = widgets.Dropdown(
    options=[('женат', 1), ('разведен', 2), ('замужем', 3), ('незамужем', 4), ('холост', 5), ('разведена', 6)],
    value=1,
    description='Семейное положение:',
)

age = widgets.IntText(
    description='Возраст: ',
    value = 0
)

work_time = widgets.IntText(
    description='Стаж: ',
    value = 0
)

children = widgets.SelectMultiple(
    options=['0', '1', '2', '3', '4'],
    value=['0'],
    rows=5,
    description='Кол-во детей',
    disabled=False
)

oklad_new = widgets.IntText(
    description='Новый оклад: ',
    value = 0
)

b3 = widgets.Button(
    description='new_oklad',
    disabled=False
)

b4 = widgets.Button(
    description='new_'
    disabled=False
)

def b1_clicked(bb):
  import datetime as dt
  global p
  p +=1
  if p >= len(Data.iloc[:,0]):
    p -= 1
  h.value = Data.iloc[p,0]
  dd = Data.iloc[p,1].split('.')
  d1.value = dt.date(int(dd[2]), int(dd[1]), int(dd[0]))
  dd = Data.iloc[p,3].split('.')
  d2.value = dt.date(int(dd[2]), int(dd[1]), int(dd[0]))
  s.value = Data.iloc[p,2]
  oklad.value = Data.iloc[p,4]
  if Data.iloc[p,5] == 'женат':
    family.value = 1
  elif Data.iloc[p,5] == 'разведен':
    family.value = 2
  elif Data.iloc[p,5] == 'замужем':
    family.value = 3
  elif Data.iloc[p,5] == 'незамужем':
    family.value = 4
  elif Data.iloc[p,5] == 'холост':
    family.value = 5
  elif Data.iloc[p,5] == 'разведена':
    family.value = 6
  else: 
    family.value = -1
  work_time.value = Data.iloc[p,7]
  age.value = Data.iloc[p,8]
  if np.isnan(Data.iloc[p,6]):
    children.value = ['0']
  else:
    children.value = [str(int(Data.iloc[p,6]))]

def b2_clicked(bb):
  import datetime as dt
  global p
  p -= 1
  if p < 0:
    p  = 0
  h.value = Data.iloc[p,0]
  dd = Data.iloc[p,1].split('.')
  d1.value =  dt.date(int(dd[2]), int(dd[1]), int(dd[0]))
  dd = Data.iloc[p,3].split('.')
  d2.value =  dt.date(int(dd[2]), int(dd[1]), int(dd[0]))
  s.value = Data.iloc[p,2]
  oklad.value = Data.iloc[p,4]
  if Data.iloc[p,5] == 'женат':
    family.value = 1
  elif Data.iloc[p,5] == 'разведен':
    family.value = 2
  elif Data.iloc[p,5] == 'замужем':
    family.value = 3
  elif Data.iloc[p,5] == 'незамужем':
    family.value = 4
  elif Data.iloc[p,5] == 'холост':
    family.value = 5
  elif Data.iloc[p,5] == 'разведена':
    family.value = 6
  else: 
    family.value = -1
  work_time.value = Data.iloc[p,7]
  age.value = Data.iloc[p,8]
  if np.isnan(Data.iloc[p,6]):
    children.value = ['0']
  else:
    children.value = [str(int(Data.iloc[p,6]))]

def b3_clicked(bb):
  global p
  Data.iloc[p,4] = oklad_new.value

def b4_clicked(bb):
  global p
  Data.iloc[p,4] = oklad_new.value

b1.on_click(b1_clicked)
b2.on_click(b2_clicked)
b3.on_click(b3_clicked)
b4.on_click(b4_clicked)


display(h,d1,d2,s,oklad,family,children,work_time,age,oklad_new,b3,b1,b2)
```
